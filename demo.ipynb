# univercmp — live demo

Cross-platform package version comparison for **APK**, **DEB**, **RPM** and **SemVer**.

Run any cell with `Shift+Enter`.

```python
from univercmp import compare, validate, PackageType
```

## Compare two versions

```python
result = compare("1.0.0-alpha", "1.0.0", kind=PackageType.SEMVER)
print(f"verdict : {result.verdict}")   # older
print(f"order   : {result.order}")     # -1
print(f"older   : {result.older}")     # True
```

## All four ecosystems

```python
cases = [
    ("1.0_rc1",              "1.0",               "apk"),
    ("1:1.0-1",              "2.0-1",              "deb"),
    ("4.18.0-513.24.1.el8_9","4.18.0-553.el8_10", "rpm"),
    ("1.0.0-beta.11",        "1.0.0-rc.1",        "semver"),
]

for src, dst, kind in cases:
    r = compare(src, dst, kind=kind)
    sym = "<" if r.older else (">" if r.newer else "=")
    print(f"[{kind:6}]  {src}  {sym}  {dst}")
```

## Validate a version string

```python
print(validate("1.0_rc1",  kind="apk"))   # True
print(validate("1.0_foo",  kind="apk"))   # False
print(validate("v1.0.0",   kind="semver")) # False — v-prefix not in spec
print(validate("2:1.0-1",  kind="deb"))   # True
```
